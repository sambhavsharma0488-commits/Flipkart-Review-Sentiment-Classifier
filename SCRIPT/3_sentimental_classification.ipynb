```python
import pandas as pd
from transformers import pipeline

df = pd.read_csv('Review_unlabelled.csv')
print(df.head())
df = df.dropna()
print(df.isnull().sum())
```

```python
classifier = pipeline(
    task="sentiment-analysis",
    model="model_sentiment",
    tokenizer="model_sentiment",
    truncation=True,
    max_length=128,
    batch_size=512
)
```

```python
print("working")
results = classifier(df["review_summary"].tolist())
print("done")
```

```python
label_li = []
score_li = []
for i in results:
    label_li.append(i["label"])
    score_li.append(i["score"])

df["sentiment"] = label_li
df["confidence_score"] = score_li
print(df.head())
```

```python
df.to_csv("Review_labelled.csv", index=False)
```
