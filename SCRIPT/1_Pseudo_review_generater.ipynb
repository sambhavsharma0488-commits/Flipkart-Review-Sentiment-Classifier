```python
import pandas as pd
from transformers import pipeline

df=pd.read_csv("Review_unlabelled.csv")
print(df.info())
print(df.head())
print(df.shape)
df=df.dropna()
```

```python
sample_df=df.sample(n=5000,random_state=42).reset_index(drop=True)
print(sample_df.head())
print(sample_df.shape)
```

```python


classifier=pipeline(
    task="sentiment-analysis",
    model="cardiffnlp/twitter-roberta-base-sentiment-latest",
    truncation=True,

)
```

```python
results=classifier.predict(sample_df['review_summary'].tolist())
```

```python
label_li=[]
score_li=[]
for i in results:
    label_li.append(i['label'])
    score_li.append(i['score'])
```

```python
sample_df['labels']=label_li
sample_df['confidence_score']=score_li
```

```python
print(sample_df.head())
print(sample_df.shape)
print(sample_df.info())
print(sample_df['labels'].value_counts())
```

```python
sample_df.to_csv("Pseudo_Review_labelled.csv",index=False)
```
